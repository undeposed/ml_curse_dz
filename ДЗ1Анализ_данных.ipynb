# Ускорение работы с pandas


Считайте датасет из файла train.csv с помощью pandas

```python
import bottleneck as bn

mean_age = bn.nanmean(df_pd["Age"].values)
std_age = bn.nanstd(df_pd["Age"].values)

print(f"Средний возраст пассажиров: {mean_age:.2f}")
print(f"Стандартное отклонение возраста: {std_age:.2f}")
```

Для каждого пассажира умножьте значение столбца Fare на 1.3, если класс его билета - 1 или 2, и на 1.1, если класс его билета - 3. Сохраните результаты как новый столбец Fare_new.

```python
import numpy as np

df_pd["Fare_new"] = np.where(
    df_pd["Pclass"].isin([1, 2]),
    df_pd["Fare"] * 1.3,
    df_pd["Fare"] * 1.1
)

df_pd[["Pclass", "Fare", "Fare_new"]].head(10)
```

# Оптимизация типов pandas


Считайте датасет из файла Housing.csv  с помощью pandas

```python
import pandas as pd

df_pd1 = pd.read_csv("Housing.csv")
df_pd1.head(10)
```

```python
print(df_pd1.describe(include="all"))


print("\nТекущие типы:")
print(df_pd1.dtypes)
```

Комментарии:

- 'price' — float64 → можно оставить float32 (max int32: 2.10000e+09, а max цена  1.330000e+07)
- 'area' — int64 → можно перевести в int16 (max площадь 16200, а max int16 = 32k)
- 'bedrooms', 'bathrooms' — int64 → int8 (значения 0–10)
- 'stories' — int64 → int8 (значения 0–10)
- 'mainroad', 'guestroom', 'basement', 'hotwaterheating',
  'airconditioning', 'prefarea' — object, но содержат 'yes'/'no' → можно заменить на bool
- 'furnishingstatus' — object, категориальный ('furnished', 'semi-furnished', 'unfurnished') → category

```python
mem_before = df_pd1.memory_usage(deep=True).sum() / 1024**2

df_opt = df_pd1.copy()

df_opt = df_opt.astype({
    "price": "float32",
    "area": "int16",
    "bedrooms": "int8",
    "bathrooms": "int8",
    "stories": "int8",
    "parking": "int8"
})

bool_cols = [
    "mainroad", "guestroom", "basement",
    "hotwaterheating", "airconditioning", "prefarea"
]
for col in bool_cols:
    df_opt[col] = df_opt[col].map({"yes": True, "no": False}).astype("bool")

df_opt["furnishingstatus"] = df_opt["furnishingstatus"].astype("category")

mem_after = df_opt.memory_usage(deep=True).sum() / 1024**2

print(f"Память до оптимизации: {mem_before:.3f} MB")
print(f"Память после оптимизации: {mem_after:.3f} MB")
print(f"Снижение потребления памяти: {(1 - mem_after / mem_before) * 100:.2f}%")


print(df_opt.dtypes)
```
